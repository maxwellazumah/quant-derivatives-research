# Options Chain Pipeline

This notebook validates a deterministic option chain, computes IV/Greek/surface features, and runs a walk-forward signal check.

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent / 'src'))

from data.synthetic import generate_synthetic_chain, generate_underlying_series
from data.schema import validate_option_schema
from options_pipeline.features import build_walk_forward_labels, compute_iv_surface_features, walk_forward_option_signal_validation
import pandas as pd

chain = generate_synthetic_chain()
validate_option_schema(chain)
features = compute_iv_surface_features(chain)
features.head()
```

```python
underlying = generate_underlying_series(periods=180)
labels = build_walk_forward_labels(underlying)
chains = [generate_synthetic_chain(as_of=row.date, spot=float(row.close), seed=100+i) for i, row in enumerate(underlying.iloc[::8].itertuples(index=False))]
features_many = compute_iv_surface_features(pd.concat(chains, ignore_index=True))
walk_forward_option_signal_validation(features_many, labels)
```
