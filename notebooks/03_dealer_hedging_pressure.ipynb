# Dealer Hedging Pressure Proxy Research

This notebook builds public-data GEX/vanna/charm-style proxies and validates them with chronological labels.

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent / 'src'))

from data.synthetic import generate_synthetic_chain, generate_underlying_series
from hedging_pressure.proxies import build_pressure_dashboard_frame, compute_dealer_pressure_proxies, walk_forward_pressure_validation
from options_pipeline.features import build_walk_forward_labels
import pandas as pd

chain = generate_synthetic_chain()
pressure = compute_dealer_pressure_proxies(chain)
dashboard = build_pressure_dashboard_frame(pressure)
dashboard.sort_values('pin_zone_score', ascending=False).head()
```

```python
underlying = generate_underlying_series(periods=180)
labels = build_walk_forward_labels(underlying)
chains = [generate_synthetic_chain(as_of=row.date, spot=float(row.close), seed=200+i) for i, row in enumerate(underlying.iloc[::10].itertuples(index=False))]
pressure_many = compute_dealer_pressure_proxies(pd.concat(chains, ignore_index=True))
walk_forward_pressure_validation(build_pressure_dashboard_frame(pressure_many), labels)
```
